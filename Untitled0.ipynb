22 - Crio uma função que pega o texto e o deslocamento pra cada letra uso o ord() pra pegar o valor dela aplico o deslocamento e o resto da divisão por 26 pra não sair do alfabeto depois volto pra letra com chr() se não for letra deixo igual. No final peço a mensagem e o deslocamento e mostro o resultado.

```python
def cifra_cesar(texto, deslocamento):
    resultado = ""

    for letra in texto:
        if letra.islower():
            nova = chr((ord(letra) - ord('a') + deslocamento) % 26 + ord('a'))
            resultado += nova
        elif letra.isupper():
            nova = chr((ord(letra) - ord('A') + deslocamento) % 26 + ord('A'))
            resultado += nova
        else:
            resultado += letra

    return resultado

mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

print(cifra_cesar(mensagem, deslocamento))
```

23 - Uso a mesma função de criptografar só que passo o deslocamento negativo. Assim em vez de avançar as letras ela volta e a mensagem fica descriptografada.

```python
def descriptografar(texto, deslocamento):
    resultado = ""

    for letra in texto:
        if letra.islower():
            letra_nova = chr((ord(letra) - ord('a') - deslocamento) % 26 + ord('a'))
            resultado += letra_nova
        elif letra.isupper():
            letra_nova = chr((ord(letra) - ord('A') - deslocamento) % 26 + ord('A'))
            resultado += letra_nova
        else:
            resultado += letra

    return resultado

mensagem = input("Digite a mensagem cifrada: ")
deslocamento = int(input("Digite o deslocamento: "))

print(descriptografar(mensagem, deslocamento))
```

24 - Tiro os pontos e o traço do CPF e confiro se tem 11 números. Calculo o primeiro dígito verificador multiplicando os 9 primeiros números por pesos decrescentes e tirando o resto da divisão por 11. Faço o mesmo pro segundo dígito usando os 10 primeiros. No final comparo se os dígitos calculados batem com os dois últimos do CPF se baterem é válido.

```python
def validar_cpf(cpf):
    cpf = cpf.replace(".", "").replace("-", "")

    if len(cpf) != 11:
        return False

    soma = 0
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)
    resto = soma % 11
    digito1 = 0 if resto < 2 else 11 - resto

    soma = 0
    for i in range(10):
        soma += int(cpf[i]) * (11 - i)
    resto = soma % 11
    digito2 = 0 if resto < 2 else 11 - resto

    if int(cpf[9]) == digito1 and int(cpf[10]) == digito2:
        return True
    return False

cpf = input("Digite o CPF: ")
if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

48 - O programa pega cada número e compara com o anterior. Se estiverem na ordem errada troca os dois de lugar. No final a lista fica ordenada do menor para o maior.

```python
numeros = [12, 2, 6, 8, 1, 3]

for i in range(1, len(numeros)):
    j = i

    while j > 0 and numeros[j] < numeros[j - 1]:
        numeros[j], numeros[j - 1] = numeros[j - 1], numeros[j]
        j -= 1

print(numeros)
```

49 - Peço um número enquanto ele for maior que zero pego o último dígito com o resto da divisão por 10 e somo no total. Depois tiro esse dígito do número com a divisão inteira por 10. Isso repete até o número chegar a zero e no final mostro a soma.

```python
n = int(input("Digite um número: "))
total = 0

while n > 0:
    ultimo = n % 10
    total = total + ultimo
    n = n // 10

print("Soma:", total)
```

50 - Peço uma frase e passo por cada letra. Se ela já tá no dicionário aumento a contagem. Se não começo com 1. No final mostro quantas vezes cada letra apareceu.

```python
texto = input("Digite uma frase: ").lower()

frequencia = {}

for letra in texto:
    if letra.isalpha():
        if letra in frequencia:
            frequencia[letra] += 1
        else:
            frequencia[letra] = 1

print(frequencia)
```
