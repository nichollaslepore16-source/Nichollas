19 - Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

# Criando os dois arrays
array1 = np.random.randint(1, 101, 20)
array2 = np.random.randint(1, 101, 20)

# Calculando o produto escalar
produto_escalar = np.dot(array1, array2)

# Calculando o tamanho dos arrays
tamanho1 = np.linalg.norm(array1)
tamanho2 = np.linalg.norm(array2)

# Calculando o ângulo
cosseno = produto_escalar / (tamanho1 * tamanho2)
angulo = np.degrees(np.arccos(cosseno))

print("Array 1:", array1)
print("Array 2:", array2)
print("Ângulo entre os arrays:", round(angulo, 2), "graus")
```

2 - Resolva a equação diferencial dy
dx = x − y, com y(0) = 1, no intervalo de 0
a 10.

```python
from scipy.integrate import solve_ivp
import numpy as np

# Eu defino a equação diferencial
def equacao(x, y):
    return x - y

# Eu resolvo de 0 até 10, começando com y(0) = 1
solucao = solve_ivp(equacao, [0, 10], [1], t_eval=np.linspace(0, 10, 50))

print("x:", solucao.t)
print("y:", solucao.y[0])
```

3 -  Encontre as raízes da função f(x) = x
4 − 3x
3 + 2 utilizando SciPy.

```python
from scipy.optimize import fsolve

# Eu defino a função
def f(x):
    return x**4 - 3*x**3 + 2

# Eu escolho alguns valores para procurar as raízes
chutes = [1, 3]
raizes = [fsolve(f, chute)[0] for chute in chutes]

print("Raízes encontradas:", raizes)
```

4 - Use a função de otimização de SciPy para maximizar f(x) = −x
2+ 4x+ 1.

```python
from scipy.optimize import minimize

# Eu defino a função
def f(x):
    return -x**2 + 4*x + 1

# Como o SciPy só tem minimize eu minimizo o negativo da função pra achar o máximo
def f_negativa(x):
    return -f(x)

resultado = minimize(f_negativa, 0)
print("Máximo encontrado em x =", resultado.x)
print("Valor máximo de f(x) =", -resultado.fun)
```

7 - Use a função de interpolação cúbica de SciPy para interpolar os dados
(0, 0),(1, 2),(2, 3),(3, 5),(4, 4).

```python
from scipy.interpolate import interp1d
import numpy as np

# Eu defino os pontos conhecidos
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

# Eu ligo os pontos com uma linha curva sem cantos
f = interp1d(x, y, kind='cubic')

# Eu descubro valores estimados entre os pontos originais
x_novos = np.array([0.5, 1.5, 2.5, 3.5])
y_novos = f(x_novos)

print("Valores estimados:", y_novos)
```

9 -  Ajuste uma curva quadrática aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8).

```python
from scipy.optimize import curve_fit
import numpy as np

# Eu defino os pontos
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

# Eu defino o formato da curva (parábola)
def parabola(x, a, b, c):
    return a*x**2 + b*x + c

# Eu ajusto a curva aos pontos
parametros, _ = curve_fit(parabola, x, y)
print("Parâmetros [a, b, c]:", parametros)
```

16 -  Resolva a equação diferencial de segunda ordem d
2y
dx2 +y = 0, com condições
iniciais y(0) = 1 e y
′
(0) = 0

```python
from scipy.integrate import solve_ivp
import numpy as np

# Eu transformo a equação de 2ª ordem em duas de 1ª ordem
# y[0] é o valor de y, y[1] é a derivada de y (y')
def equacao(t, y):
    dy0 = y[1]
    dy1 = -y[0]
    return [dy0, dy1]

# Eu resolvo com y(0)=1 e y'(0)=0, no intervalo de 0 a 10
solucao = solve_ivp(equacao, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 50))

print("t:", solucao.t)
print("y:", solucao.y[0])
```

```python
import pandas as pd

# Eu crio um DataFrame com uma coluna que tem valores repetidos
dados = {'Produto': ['Camisa', 'Calça', 'Tênis', 'Boné', 'Meia'],
         'Categoria': ['Roupa', 'Roupa', 'Calçado', 'Acessório', 'Acessório'],
         'Preço': [50, 120, 200, 40, 15]}
df = pd.DataFrame(dados)

# Eu verifico, para cada coluna, se ela tem valores duplicados
for coluna in df.columns:
    tem_duplicado = df[coluna].duplicated().any()
    print(coluna, "-", "tem duplicado" if tem_duplicado else "sem duplicado")
```

8 - Plote um gráfico de linha com intervalos de confiança (sombramento) para
a função y = sin(x).

```python
import matplotlib.pyplot as plt
import numpy as np

# Eu gero 100 valores de x entre 0 e um ciclo completo (2pi), e calculo o seno de cada um
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Eu marco uma faixa de 0.2 acima e abaixo da linha, simulando uma margem de erro
limite_inferior = y - 0.2
limite_superior = y + 0.2

plt.plot(x, y, label="sin(x)")

# Eu preencho a área entre os dois limites, criando o efeito de sombreamento
plt.fill_between(
    x,
    limite_inferior,
    limite_superior,
    alpha=0.3,      # controla o quão transparente fica o preenchimento
    label="Intervalo de confiança"
)

plt.title("Função seno com intervalo de confiança")
plt.xlabel("X")
plt.ylabel("Y")

# Eu adiciono a legenda, usando os textos definidos em label
plt.legend()

plt.show()
```

10 - Plote três gráficos de linha em um único gráfico, cada um com uma cor e
estilo de linha diferentes.

```python
import matplotlib.pyplot as plt
import numpy as np

# Eu gero os dados para as três linhas
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.plot(x, y1, color='blue', linestyle='-', label='sin(x)')
plt.plot(x, y2, color='red', linestyle='--', label='cos(x)')
plt.plot(x, y3, color='green', linestyle=':', label='sin(x)*cos(x)')

plt.title('Três Linhas no Mesmo Gráfico')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.legend()
plt.show()
```

11 - Crie um gráfico de barras empilhadas para três categorias com três grupos
de valores.

```python
import matplotlib.pyplot as plt

# Eu defino as categorias e os três grupos de valores
categorias = ['A', 'B', 'C']
grupo1 = [10, 15, 7]
grupo2 = [5, 8, 12]
grupo3 = [3, 6, 9]

plt.bar(categorias, grupo1, label='Grupo 1')
plt.bar(categorias, grupo2, bottom=grupo1, label='Grupo 2')

# Eu somo os dois primeiros grupos para saber onde o terceiro deve começar
soma_anterior = [g1 + g2 for g1, g2 in zip(grupo1, grupo2)]
plt.bar(categorias, grupo3, bottom=soma_anterior, label='Grupo 3')

plt.title('Gráfico de Barras Empilhadas')
plt.xlabel('Categorias')
plt.ylabel('Valores')
plt.legend()
plt.show()
```
